### **Đặc trưng Histogram ảnh màu**
**Histogram ảnh màu** là biểu đồ thống kê tần suất xuất hiện của các giá trị màu trong ảnh. Đối với ảnh màu, histogram thường được chia thành ba kênh màu: *đỏ, xanh lá cây và xanh dương*. Mỗi kênh màu có một histogram riêng của nó, thể hiện sự phân phối của các cường độ màu trong kênh đó. Đặc trưng histogram ảnh màu cung cấp thông tin về sự phân bố màu sắc trong ảnh, giúp phát hiện các đặc điểm quan trọng như sự phân biệt màu sắc và sự cân bằng màu sắc.

```python
import cv2
import matplotlib.pyplot as plt

# Đọc đường dẫn của ảnh nếu cần thiết
img_path = '/content/drive/MyDrive/cv/0. Dataset/0001_001.jpg'
img_read = cv2.imread(img_path)

# Chuyển ảnh sang không gian Xám (Gray-scale)
gray_img = cv2.cvtColor(img_read, cv2.COLOR_BGR2GRAY)

# Hiển thị hình ảnh
plt.figure(figsize=(10, 5)) # KHỞI TẠO FIGURE GRID 10x5

plt.subplot(1, 2, 1) # Row: 1 - Column: 2 - Index: 1
plt.imshow(cv2.cvtColor(img_read, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for matplotlib
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2) # Row: 1 - Column: 2 - Index: 1
plt.imshow(gray_img, cmap='gray')
plt.title('Gray Image')
plt.axis("off")

plt.show()
```

```python
image = img_read

# Cách 1 sử dụng OpenCV2
hist_b  = cv2.calcHist(image, [0], None, [256], [0, 256])
hist_g = cv2.calcHist(image, [1], None, [256], [0, 256])
hist_r  = cv2.calcHist(image, [2], None, [256], [0, 256])

# Vẽ biểu đồ
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(hist_b, color='b')
plt.title('Histogram - Blue')
plt.xlim([0, 256])

plt.subplot(1, 3, 2)
plt.plot(hist_g, color='g')
plt.title('Histogram - Green')
plt.xlim([0, 256])

plt.subplot(1, 3, 3)
plt.plot(hist_r, color='r')
plt.title('Histogram - Red')
plt.xlim([0, 256])

plt.tight_layout()
plt.show()
```

```python
import numpy as np

# Cách 2 sử dụng numpy
hist_b , bins_ = np.histogram(img_read[:,:,0].ravel(), bins = 256, range = [0,256])
hist_g , bins_ = np.histogram(img_read[:,:,1].ravel(), bins = 256, range = [0,256])
hist_r , bins_ = np.histogram(img_read[:,:,2].ravel(), bins = 256, range = [0,256])

# Vẽ biểu đồ
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(hist_b, color='b')
plt.title('Histogram - Blue')
plt.xlim([0, 256])

plt.subplot(1, 3, 2)
plt.plot(hist_g, color='g')
plt.title('Histogram - Green')
plt.xlim([0, 256])

plt.subplot(1, 3, 3)
plt.plot(hist_r, color='r')
plt.title('Histogram - Red')
plt.xlim([0, 256])

plt.tight_layout()
plt.show()
```

```python
# Cách 3 dùng vòng lặp
height, width, channel = image.shape

# Khởi tạo histogram: 256 mức sáng
hist_b = np.zeros(256, dtype=int)
hist_g = np.zeros(256, dtype=int)
hist_r = np.zeros(256, dtype=int)

for y in range(height):
    for x in range(width):
        b,g,r = image[y, x]
        hist_b[b] += 1
        hist_g[g] += 1
        hist_r[r] += 1

# Vẽ biểu đồ
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(hist_b, color='b')
plt.title('Histogram - Blue')
plt.xlim([0, 256])

plt.subplot(1, 3, 2)
plt.plot(hist_g, color='g')
plt.title('Histogram - Green')
plt.xlim([0, 256])

plt.subplot(1, 3, 3)
plt.plot(hist_r, color='r')
plt.xlim([0, 256])

plt.tight_layout()
plt.show()
```

## **3.2. Tính toán Histogram và Ranking**

```python
# Khởi tạo hành tính Histogram cho ảnh
def calculate_histogram(image_path):
    image_read = cv2.imread(image_path)

    gray = cv2.cvtColor(image_read, cv2.COLOR_BGR2GRAY)

    # Tính histogram ảnh grayscale (1 kênh, 256 bins, giá trị từ 0 đến 255)
    hist = cv2.calcHist([gray], [0], None, [256], [0,256])

    # Normalize để chuẩn hóa histogram (giá trị về cùng 1 thang đo)
    hist = cv2.normalize(hist, hist).flatten()
    return hist
```

```python
import os

path = '/content/drive/MyDrive/cv/0. Dataset'

image_paths = []
for root, dirs, files in os.walk(path):
    for file in files:
        if file.lower().endswith(".jpg"):
            image_paths.append(os.path.join(root, file))

histograms = []
for image_path in image_paths:
    hist = calculate_histogram(image_path)
    histograms.append(hist)
```

```python

# Tính toán Histogram cho ảnh query
query_image_path = '/content/drive/MyDrive/cv/0. Dataset/0001_001.jpg'
histogram_image_query = calculate_histogram(query_image_path)
```

```python
import matplotlib.pyplot as plt
def compare_histograms_and_show(query_hist, dataset_hists, image_paths, method=cv2.HISTCMP_INTERSECT, top_k=10):
    results = []

    for hist, path in zip(dataset_hists, image_paths):
        score = cv2.compareHist(query_hist, hist, method)
        results.append((path, score))

    results.sort(key=lambda x: x[1], reverse=True)

    # Lấy top K kết quả (ảnh giống nhất)
    top_results = results[:top_k]

    # Hiển thị ảnh truy vấn + top ảnh tương tự
    plt.figure(figsize=(22, 5))

    # Ảnh truy vấn là subplot đầu tiên
    plt.subplot(1, top_k + 1, 1)
    query_img = cv2.imread(query_image_path)
    query_img_rgb = cv2.cvtColor(query_img, cv2.COLOR_BGR2RGB)
    plt.imshow(query_img_rgb)
    plt.axis('off')
    plt.title("Query")

    # Các ảnh giống nhất
    for i, (path, score) in enumerate(top_results):
        img = cv2.imread(path)
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        plt.subplot(1, top_k + 1, i + 2)
        plt.imshow(img_rgb)
        plt.axis('off')
        plt.title(f"{score:.2f}")

    plt.tight_layout()
    plt.show()

compare_histograms_and_show(
    histogram_image_query,
    histograms,
    image_paths,
    top_k=10
)
```

## **3.3. (Add-on) Tính toán Histogram của ảnh màu và Ranking**

```python
def calculate_color_histogram(image_path):
    image_read = cv2.imread(image_path)
    if image_read is None:
        return None
    hist = cv2.calcHist([image_read], [0, 1, 2], None, [8, 8, 8], [0, 256, 0, 256, 0, 256])
    hist = cv2.normalize(hist, hist).flatten()
    return hist


def compare_color_histograms_and_show(query_hist, dataset_hists, image_paths,
                                      method=cv2.HISTCMP_CORREL, top_k=10):

    results = []

    for hist, path in zip(dataset_hists, image_paths):
        score = cv2.compareHist(query_hist, hist, method)
        results.append((path, score))

    results.sort(key=lambda x: x[1], reverse=True)

    top_results = results[:top_k]

    plt.figure(figsize=(22,5))

    # Query image
    plt.subplot(1, top_k+1, 1)
    query_img = cv2.imread(query_image_path)
    query_img_rgb = cv2.cvtColor(query_img, cv2.COLOR_BGR2RGB)
    plt.imshow(query_img_rgb)
    plt.axis('off')
    plt.title("Query")

    # Top K images
    for i, (path, score) in enumerate(top_results):
        img = cv2.imread(path)
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        plt.subplot(1, top_k+1, i+2)
        plt.imshow(img_rgb)
        plt.axis('off')
        plt.title(f"{score:.2f}")

    plt.tight_layout()
    plt.show()

color_histograms = []
for image_path in image_paths:
    hist = calculate_color_histogram(image_path)
    if hist is not None:
        color_histograms.append(hist)

query_image_path = '/content/drive/MyDrive/cv/0. Dataset/0001_001.jpg'
query_color_hist = calculate_color_histogram(query_image_path)



compare_color_histograms_and_show(
    query_color_hist,
    color_histograms,
    image_paths,
    method=cv2.HISTCMP_CORREL,
    top_k=10
)
```

# **# 05: Cân bằng xám**

```python
import cv2
import matplotlib.pyplot as plt

image_path = '/content/drive/MyDrive/cv/0. Dataset/0002_002.jpg'
image_gray = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

histogram_image_gray = cv2.calcHist([image_gray], [0], None, [256], [0,256])

equalization_image = cv2.equalizeHist(image_gray)

histogram_equalize = cv2.calcHist([equalization_image], [0], None, [256], [0,256])

plt.figure(figsize=(12, 6))

plt.subplot(2, 2, 1)
plt.imshow(image_gray, cmap='gray')
plt.title('Ảnh gốc')

plt.subplot(2, 2, 2)
plt.plot(histogram_image_gray, color='black')
plt.title('Histogram gốc')

plt.subplot(2, 2, 3)
plt.imshow(equalization_image, cmap='gray')
plt.title('Ảnh sau cân bằng')

plt.subplot(2, 2, 4)
plt.plot(histogram_equalize, color='black')
plt.title('Histogram sau cân bằng')

plt.tight_layout()
plt.show()
```

```python
import numpy as np

# Add-on thực hiện thủ công trên ma trận 3x3 và hiển thị ma trận sau khi cân bằng xám
image = np.array([
    [52, 55, 61],
    [66, 70, 61],
    [64, 73, 52]
], dtype=np.uint8)

# 1. Tính histogram (256 mức xám)
hist = np.zeros(256, dtype=int)

for value in image.flatten():
    hist[value] += 1

print("\nHistogram (các giá trị xuất hiện):")
for i in range(256):
    if hist[i] > 0:
        print(f"{i}: {hist[i]}")

# 2. Tính CDF
cdf = hist.cumsum()

# 3. Chuẩn hóa CDF
cdf_min = cdf[cdf > 0].min()
N = image.size

mapping = np.round((cdf - cdf_min) / (N - cdf_min) * 255).astype(np.uint8)

# 4. Áp dụng mapping
equalized_image = mapping[image]

print("\nMa trận sau cân bằng xám:\n", equalized_image)
```
