## *2.2. Bài tập 1B*
- Viết hàm tự chuyển đổi từ ảnh 3 kênh màu sang ảnh thang mức xám

```python
import cv2
import numpy as np

def rgb_to_grayscale(image):
    # Lấy kích thước của ảnh
    height, width, channels = image.shape

    # Tạo một mảng mới để lưu ảnh grayscale
    grayscale_image = np.zeros((height, width), dtype=np.uint8)

    # Duyệt qua từng pixel trong ảnh
    for i in range(height):
        for j in range(width):
            B = image[i, j, 0]
            G = image[i, j, 1]
            R = image[i, j, 2]

            gray = 0.333 * B + 0.333 * G + 0.333 * R
            grayscale_image[i, j] = int(gray)

    return grayscale_image

# Đọc ảnh RGB từ file
image_path = '/content/drive/MyDrive/cv/Images/Lena.png'
image = cv2.imread (image_path)

# Chuyển đổi ảnh sang thang mức xám (grayscale)
gray_image = rgb_to_grayscale(image)
#cv2.save

# Hiển thị ảnh thang mức xám
plt.imshow(gray_image, cmap='gray')
plt.show()
```
